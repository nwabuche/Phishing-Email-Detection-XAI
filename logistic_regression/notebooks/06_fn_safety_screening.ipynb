# Logistic Regression — False Negative Safety Screening

## Purpose

This notebook evaluates whether the counterfactual feature changes in the approved synthetic legitimate emails could weaken the detection of phishing emails. The screening uses the original training data and the fixed Term Frequency-Inverse Document Frequency representation. It does not use the held-out test set to determine safety.

The investigation follows the False Positive error analysis informed by **SHapley Additive exPlanations (SHAP)** and the counterfactual-guided generation of legitimate email variants. SHAP provides the preceding explanation of model errors; the safety assessment itself is performed through controlled feature-reduction tests.

## Method

The analysis has two levels. First, individual changed features are tested on correctly detected phishing examples from five training-only validation folds. Second, the complete feature-change pattern associated with each synthetic email is tested on the same validation setting. A feature is accepted only when it induces no additional False Negative predictions. A synthetic email must pass both the individual-feature and combined-pattern tests.

The official experiment screened **34** approved variants, examined **93** unique changed features, and identified **9** variants that passed the safety criteria. These are recorded experimental checkpoints, not newly measured results until the code is executed successfully.

```python
# =========================================================================================================

# STEP 12G: COUNTERFACTUAL FN-SAFETY SCREENING

# TOKEN-LEVEL + SYNTHETIC-PATTERN STRESS TEST

# TRAINING-ONLY VALIDATION

# =========================================================================================================

#

# PHASE 3 — COUNTERFACTUAL-GUIDED SYNTHETIC AUGMENTATION

#

# PURPOSE

# -------

# Before using synthetic legitimate emails for model retraining,

# determine whether their counterfactual token changes could weaken

# phishing detection and create false negatives.

#

# TWO LEVELS OF SAFETY TEST

# -------------------------

#

# LEVEL 1 — TOKEN SAFETY

# Each changed feature is tested independently on correctly detected

# phishing validation emails.

#

# LEVEL 2 — SYNTHETIC PATTERN SAFETY

# The complete combination of changes belonging to each synthetic email

# is tested on correctly detected phishing validation emails.

#

#

# STRICT SAFETY RULE

# ------------------

#

# A token is FN-safe only if:

#

#       induced FN flips = 0

#

# A synthetic email is FN-safe only if:

#

#       all its changed tokens are individually FN-safe

#

#       AND

#

#       its complete counterfactual pattern causes 0 induced FN flips

#

#

# IMPORTANT

# ---------

#

# - No synthetic email is added to training here.

# - No augmented Logistic Regression model is trained.

# - TF-IDF vocabulary is NOT refitted.

# - Threshold remains 0.50.

# - No coefficient is manually modified.

#

# =========================================================================================================





```

```python
# =========================================================================================================

# 1. IMPORTS

# =========================================================================================================



import os

import json

import hashlib



import numpy as np

import pandas as pd



from scipy import sparse



from sklearn.linear_model import LogisticRegression

from sklearn.model_selection import StratifiedKFold





print("=" * 125)

print("STEP 12G: COUNTERFACTUAL FN-SAFETY SCREENING")

print("TOKEN-LEVEL + SYNTHETIC-PATTERN STRESS TEST")

print("=" * 125)





```

```python
# =========================================================================================================

# 2. VERIFY REQUIRED OBJECTS

# =========================================================================================================



required_objects_12g = [



    "X_train_tfidf",

    "y_train",



    "tfidf_vectorizer",



    "synthetic_final_approved_12f",

    "synthetic_refined_change_log_12f"

]





missing_objects_12g = [



    obj



    for obj in required_objects_12g



    if obj not in globals()

]





if missing_objects_12g:



    raise RuntimeError(



        "\nRequired Phase 3 objects are missing:\n"



        f"{missing_objects_12g}\n\n"



        "Run the verified Step 12F first."

    )





```

```python
# =========================================================================================================

# 3. FIXED CONFIGURATION

# =========================================================================================================



RANDOM_STATE_12G = 42



THRESHOLD_12G = 0.50



GLOBAL_C_12G = 1.0



MAX_ITER_12G = 1000





EXPECTED_TRAIN_12G = 42861



EXPECTED_FEATURES_12G = 10000



EXPECTED_APPROVED_SYNTHETIC_12G = 34





MAX_ALLOWED_INDUCED_FN_FLIPS_12G = 0





PROJECT_ROOT_12G = (
    "/content/drive/MyDrive/Logistic_Regression_SHAP_Guided_Phishing_Detection"
    if os.path.isdir("/content/drive/MyDrive")
    else os.path.join(os.getcwd(), "logistic_regression", "results", "fn_safety")
)





STEP12G_FOLDER = os.path.join(

    PROJECT_ROOT_12G,

    "Step12G_Counterfactual_FN_Safety_Screening"

)





os.makedirs(

    STEP12G_FOLDER,

    exist_ok=True

)





```

```python
# =========================================================================================================

# 4. VERIFY ORIGINAL TRAINING ENVIRONMENT

# =========================================================================================================



y_train_12g = (



    np.asarray(

        y_train

    )



    .astype(int)



    .reshape(-1)

)





assert X_train_tfidf.shape == (

    EXPECTED_TRAIN_12G,

    EXPECTED_FEATURES_12G

)





assert len(

    y_train_12g

) == EXPECTED_TRAIN_12G





feature_names_12g = np.asarray(

    tfidf_vectorizer.get_feature_names_out()

)





assert len(

    feature_names_12g

) == EXPECTED_FEATURES_12G





locked_vocab_12g = (

    tfidf_vectorizer.vocabulary_

)





print("\nOriginal training environment verified.")



print(

    "Training matrix:",

    X_train_tfidf.shape

)



print(

    "TF-IDF features:",

    len(

        feature_names_12g

    )

)





```

```python
# =========================================================================================================

# 5. VERIFY APPROVED SYNTHETIC DATASET

# =========================================================================================================



approved_synthetic_12g = (

    synthetic_final_approved_12f.copy()

)





assert len(

    approved_synthetic_12g

) == EXPECTED_APPROVED_SYNTHETIC_12G





assert (

    approved_synthetic_12g[

        "label"

    ] == 0

).all()





approved_ids_12g = (



    approved_synthetic_12g[

        "Synthetic_ID"

    ]



    .astype(int)



    .tolist()

)





print(

    "\nApproved Step 12F synthetic emails:",

    len(

        approved_synthetic_12g

    )

)





```

```python
# =========================================================================================================

# 6. RECOVER VALIDATED CHANGE LOG

# =========================================================================================================



change_log_12g = (



    synthetic_refined_change_log_12f[



        synthetic_refined_change_log_12f[

            "Synthetic_ID"

        ].isin(

            approved_ids_12g

        )



    ]



    .copy()

)





change_log_12g = (



    change_log_12g[



        (

            change_log_12g[

                "Modified_Occurrences"

            ] > 0

        )



        &



        (

            change_log_12g[

                "Change_Satisfied"

            ] == True

        )



    ]



    .copy()

    .reset_index(drop=True)

)





assert len(

    change_log_12g

) > 0





print(

    "Validated feature-change records:",

    len(

        change_log_12g

    )

)





```

```python
# =========================================================================================================

# 7. RESOLVE TF-IDF INDICES

# =========================================================================================================



resolved_feature_indices_12g = []





for _, row in change_log_12g.iterrows():





    feature = str(

        row[

            "Feature"

        ]

    )





    if feature not in locked_vocab_12g:



        raise RuntimeError(

            f"Counterfactual feature '{feature}' "

            "is missing from the locked TF-IDF vocabulary."

        )





    resolved_feature_indices_12g.append(



        int(

            locked_vocab_12g[

                feature

            ]

        )

    )





change_log_12g[

    "TFIDF_Index"

] = resolved_feature_indices_12g





```

```python
# =========================================================================================================

# 8. BUILD TOKEN-LEVEL STRESS CONFIGURATION

# =========================================================================================================



token_stress_config_12g = (



    change_log_12g



    .groupby(

        [

            "Feature",

            "TFIDF_Index"

        ],

        as_index=False

    )



    .agg({



        "Actual_Fraction_Removed":

            "max",



        "Synthetic_ID":

            "nunique",



        "Source_FP_Number":

            "nunique"

    })



    .rename(

        columns={



            "Actual_Fraction_Removed":

                "Stress_Reduction_Fraction",



            "Synthetic_ID":

                "Synthetic_Email_Frequency",



            "Source_FP_Number":

                "Source_FP_Frequency"

        }

    )

)





print(

    "\nUnique counterfactual tokens to FN-stress-test:",

    len(

        token_stress_config_12g

    )

)





```

```python
# =========================================================================================================

# 9. HELPER — APPLY FEATURE REDUCTIONS

# =========================================================================================================



def apply_feature_reduction_12g(

    X,

    feature_fraction_map

):



    scale_vector = np.ones(

        X.shape[1],

        dtype=np.float64

    )





    for feature_index, fraction_removed in feature_fraction_map.items():





        fraction_removed = float(

            fraction_removed

        )





        fraction_removed = min(

            1.0,

            max(

                0.0,

                fraction_removed

            )

        )





        scale_vector[

            int(

                feature_index

            )

        ] = (

            1.0

            -

            fraction_removed

        )





    return sparse.csr_matrix(

        X.multiply(

            scale_vector

        )

    )





```

```python
# =========================================================================================================

# 10. FIXED 5-FOLD TRAINING-ONLY CV

# =========================================================================================================



cv_12g = StratifiedKFold(



    n_splits=5,



    shuffle=True,



    random_state=RANDOM_STATE_12G

)





fold_splits_12g = list(



    cv_12g.split(

        X_train_tfidf,

        y_train_12g

    )

)





```

```python
# =========================================================================================================

# 11. BUILD PHISHING SAFETY ENVIRONMENT

# =========================================================================================================



fold_environment_12g = []





print(

    "\n" + "=" * 125

)



print(

    "BUILDING TRAINING-ONLY PHISHING SAFETY ENVIRONMENT"

)



print(

    "=" * 125

)





for fold_id, (

    train_idx,

    val_idx

) in enumerate(

    fold_splits_12g,

    start=1

):





    X_fold_train = (

        X_train_tfidf[

            train_idx

        ]

    )





    X_fold_val = (

        X_train_tfidf[

            val_idx

        ]

    )





    y_fold_train = (

        y_train_12g[

            train_idx

        ]

    )





    y_fold_val = (

        y_train_12g[

            val_idx

        ]

    )





    fold_model = LogisticRegression(



        C=GLOBAL_C_12G,



        random_state=RANDOM_STATE_12G,



        max_iter=MAX_ITER_12G

    )





    fold_model.fit(

        X_fold_train,

        y_fold_train

    )





    baseline_probability = (

        fold_model.predict_proba(

            X_fold_val

        )[:, 1]

    )





    baseline_prediction = (



        baseline_probability

        >=

        THRESHOLD_12G



    ).astype(int)





    correct_phishing_mask = (



        (y_fold_val == 1)



        &



        (baseline_prediction == 1)

    )





    correct_phishing_positions = np.where(

        correct_phishing_mask

    )[0]





    X_correct_phishing = (

        X_fold_val[

            correct_phishing_positions

        ]

    )





    baseline_correct_phishing_prob = (

        baseline_probability[

            correct_phishing_positions

        ]

    )





    fold_environment_12g.append({



        "Fold":

            fold_id,



        "Model":

            fold_model,



        "X_Correct_Phishing":

            X_correct_phishing,



        "Baseline_Probabilities":

            baseline_correct_phishing_prob,



        "Correct_Phishing_Count":

            len(

                correct_phishing_positions

            )

    })





    print(

        f"Fold {fold_id}: "

        f"correct phishing available for safety stress test = "

        f"{len(correct_phishing_positions)}"

    )





```

```python
# =========================================================================================================

# 12. LEVEL 1 — TOKEN FN-SAFETY TEST

# =========================================================================================================



token_fold_records_12g = []





print(

    "\n" + "=" * 125

)



print(

    "LEVEL 1: TOKEN-LEVEL COUNTERFACTUAL FN-SAFETY TEST"

)



print(

    "=" * 125

)





for _, token_row in token_stress_config_12g.iterrows():





    feature = str(

        token_row[

            "Feature"

        ]

    )





    feature_index = int(

        token_row[

            "TFIDF_Index"

        ]

    )





    reduction_fraction = float(

        token_row[

            "Stress_Reduction_Fraction"

        ]

    )





    for fold_info in fold_environment_12g:





        fold_id = int(

            fold_info[

                "Fold"

            ]

        )





        fold_model = (

            fold_info[

                "Model"

            ]

        )





        X_correct_phishing = (

            fold_info[

                "X_Correct_Phishing"

            ]

        )





        baseline_probs = (

            fold_info[

                "Baseline_Probabilities"

            ]

        )





        feature_present_mask = np.asarray(



            (

                X_correct_phishing[

                    :,

                    feature_index

                ]

                >

                0

            )

            .todense()



        ).reshape(-1)





        affected_count = int(

            feature_present_mask.sum()

        )





        if affected_count == 0:





            token_fold_records_12g.append({



                "Feature":

                    feature,



                "TFIDF_Index":

                    feature_index,



                "Reduction_Fraction":

                    reduction_fraction,



                "Fold":

                    fold_id,



                "Affected_Phishing_Count":

                    0,



                "Induced_FN_Flips":

                    0,



                "Induced_FN_Rate_Affected":

                    0.0,



                "Mean_Probability_Change":

                    0.0,



                "Minimum_Probability_After":

                    np.nan

            })





            continue





        X_affected = (

            X_correct_phishing[

                feature_present_mask

            ]

        )





        baseline_affected_prob = (

            baseline_probs[

                feature_present_mask

            ]

        )





        X_modified = (

            apply_feature_reduction_12g(



                X_affected,



                {

                    feature_index:

                        reduction_fraction

                }

            )

        )





        modified_probability = (

            fold_model.predict_proba(

                X_modified

            )[:, 1]

        )





        modified_prediction = (



            modified_probability

            >=

            THRESHOLD_12G



        ).astype(int)





        induced_fn_flips = int(

            (

                modified_prediction == 0

            ).sum()

        )





        induced_fn_rate = (

            induced_fn_flips

            /

            affected_count

        )





        probability_change = (

            modified_probability

            -

            baseline_affected_prob

        )





        token_fold_records_12g.append({



            "Feature":

                feature,



            "TFIDF_Index":

                feature_index,



            "Reduction_Fraction":

                reduction_fraction,



            "Fold":

                fold_id,



            "Affected_Phishing_Count":

                affected_count,



            "Induced_FN_Flips":

                induced_fn_flips,



            "Induced_FN_Rate_Affected":

                induced_fn_rate,



            "Mean_Probability_Change":

                float(

                    probability_change.mean()

                ),



            "Minimum_Probability_After":

                float(

                    modified_probability.min()

                )

        })





token_fold_df_12g = pd.DataFrame(

    token_fold_records_12g

)





```

```python
# =========================================================================================================

# 13. AGGREGATE TOKEN SAFETY

# =========================================================================================================



token_safety_12g = (



    token_fold_df_12g



    .groupby(

        [

            "Feature",

            "TFIDF_Index",

            "Reduction_Fraction"

        ],

        as_index=False

    )



    .agg({



        "Affected_Phishing_Count":

            "sum",



        "Induced_FN_Flips":

            "sum",



        "Mean_Probability_Change":

            "mean",



        "Minimum_Probability_After":

            "min"

    })

)





token_safety_12g[

    "Induced_FN_Rate"

] = np.where(



    token_safety_12g[

        "Affected_Phishing_Count"

    ] > 0,



    (

        token_safety_12g[

            "Induced_FN_Flips"

        ]



        /



        token_safety_12g[

            "Affected_Phishing_Count"

        ]

    ),



    0.0

)





token_safety_12g[

    "FN_Safe_Token"

] = (



    token_safety_12g[

        "Induced_FN_Flips"

    ]



    <=



    MAX_ALLOWED_INDUCED_FN_FLIPS_12G

)





token_safety_12g = (



    token_safety_12g



    .merge(



        token_stress_config_12g[

            [

                "Feature",

                "TFIDF_Index",

                "Synthetic_Email_Frequency",

                "Source_FP_Frequency"

            ]

        ],



        on=[

            "Feature",

            "TFIDF_Index"

        ],



        how="left"

    )

)





safe_token_count_12g = int(

    token_safety_12g[

        "FN_Safe_Token"

    ].sum()

)





risky_token_count_12g = int(

    (

        token_safety_12g[

            "FN_Safe_Token"

        ] == False

    ).sum()

)





print(

    "\nToken FN-safety results:"

)



print(

    "Safe tokens:",

    safe_token_count_12g

)



print(

    "Risky tokens:",

    risky_token_count_12g

)





```

```python
# =========================================================================================================

# 14. BUILD SAFE / RISKY TOKEN SETS

# =========================================================================================================



safe_token_set_12g = set(



    token_safety_12g[



        token_safety_12g[

            "FN_Safe_Token"

        ] == True



    ][

        "Feature"

    ]

    .astype(str)

    .tolist()

)





risky_token_set_12g = set(



    token_safety_12g[



        token_safety_12g[

            "FN_Safe_Token"

        ] == False



    ][

        "Feature"

    ]

    .astype(str)

    .tolist()

)





```

```python
# =========================================================================================================

# 15. BUILD SYNTHETIC PATTERN DEFINITIONS

# =========================================================================================================



synthetic_pattern_records_12g = []





for synthetic_id in approved_ids_12g:





    sample_changes = (



        change_log_12g[



            change_log_12g[

                "Synthetic_ID"

            ] == synthetic_id



        ]



        .copy()

    )





    feature_fraction_map = {}



    changed_features = []





    for _, change_row in sample_changes.iterrows():





        feature = str(

            change_row[

                "Feature"

            ]

        )





        feature_index = int(

            change_row[

                "TFIDF_Index"

            ]

        )





        fraction = float(

            change_row[

                "Actual_Fraction_Removed"

            ]

        )





        if feature_index in feature_fraction_map:



            feature_fraction_map[

                feature_index

            ] = max(



                feature_fraction_map[

                    feature_index

                ],



                fraction

            )



        else:



            feature_fraction_map[

                feature_index

            ] = fraction





        changed_features.append(

            feature

        )





    unique_changed_features = sorted(

        set(

            changed_features

        )

    )





    risky_features_in_sample = sorted(



        set(

            unique_changed_features

        )



        &



        risky_token_set_12g

    )





    all_tokens_individually_safe = bool(

        len(

            risky_features_in_sample

        ) == 0

    )





    synthetic_pattern_records_12g.append({



        "Synthetic_ID":

            synthetic_id,



        "Feature_Fraction_Map":

            feature_fraction_map,



        "Changed_Features":

            unique_changed_features,



        "Changed_Feature_Count":

            len(

                unique_changed_features

            ),



        "All_Tokens_Individually_Safe":

            all_tokens_individually_safe,



        "Risky_Features":

            risky_features_in_sample

    })





```

```python
# =========================================================================================================

# 16. LEVEL 2 — COMPLETE PATTERN FN-SAFETY TEST

# =========================================================================================================



sample_fold_records_12g = []





print(

    "\n" + "=" * 125

)



print(

    "LEVEL 2: COMPLETE SYNTHETIC-PATTERN FN-SAFETY TEST"

)



print(

    "=" * 125

)





for sample_info in synthetic_pattern_records_12g:





    synthetic_id = int(

        sample_info[

            "Synthetic_ID"

        ]

    )





    feature_fraction_map = (

        sample_info[

            "Feature_Fraction_Map"

        ]

    )





    changed_feature_indices = list(

        feature_fraction_map.keys()

    )





    for fold_info in fold_environment_12g:





        fold_id = int(

            fold_info[

                "Fold"

            ]

        )





        fold_model = (

            fold_info[

                "Model"

            ]

        )





        X_correct_phishing = (

            fold_info[

                "X_Correct_Phishing"

            ]

        )





        baseline_probs = (

            fold_info[

                "Baseline_Probabilities"

            ]

        )





        if len(

            changed_feature_indices

        ) == 0:



            affected_mask = np.zeros(

                X_correct_phishing.shape[0],

                dtype=bool

            )



        else:



            affected_count_matrix = (



                X_correct_phishing[

                    :,

                    changed_feature_indices

                ]

                >

                0

            )





            affected_mask = np.asarray(



                affected_count_matrix.sum(

                    axis=1

                )



            ).reshape(-1) > 0





        affected_count = int(

            affected_mask.sum()

        )





        if affected_count == 0:





            sample_fold_records_12g.append({



                "Synthetic_ID":

                    synthetic_id,



                "Fold":

                    fold_id,



                "Affected_Phishing_Count":

                    0,



                "Induced_FN_Flips":

                    0,



                "Induced_FN_Rate":

                    0.0,



                "Mean_Probability_Change":

                    0.0,



                "Minimum_Probability_After":

                    np.nan

            })





            continue





        X_affected = (

            X_correct_phishing[

                affected_mask

            ]

        )





        baseline_affected_probs = (

            baseline_probs[

                affected_mask

            ]

        )





        X_modified = (

            apply_feature_reduction_12g(



                X_affected,



                feature_fraction_map

            )

        )





        modified_probability = (

            fold_model.predict_proba(

                X_modified

            )[:, 1]

        )





        modified_prediction = (



            modified_probability

            >=

            THRESHOLD_12G



        ).astype(int)





        induced_flips = int(

            (

                modified_prediction == 0

            ).sum()

        )





        induced_rate = (

            induced_flips

            /

            affected_count

        )





        probability_change = (

            modified_probability

            -

            baseline_affected_probs

        )





        sample_fold_records_12g.append({



            "Synthetic_ID":

                synthetic_id,



            "Fold":

                fold_id,



            "Affected_Phishing_Count":

                affected_count,



            "Induced_FN_Flips":

                induced_flips,



            "Induced_FN_Rate":

                induced_rate,



            "Mean_Probability_Change":

                float(

                    probability_change.mean()

                ),



            "Minimum_Probability_After":

                float(

                    modified_probability.min()

                )

        })





sample_fold_df_12g = pd.DataFrame(

    sample_fold_records_12g

)





```

```python
# =========================================================================================================

# 17. AGGREGATE SAMPLE SAFETY

# =========================================================================================================



sample_safety_12g = (



    sample_fold_df_12g



    .groupby(

        "Synthetic_ID",

        as_index=False

    )



    .agg({



        "Affected_Phishing_Count":

            "sum",



        "Induced_FN_Flips":

            "sum",



        "Mean_Probability_Change":

            "mean",



        "Minimum_Probability_After":

            "min"

    })

)





sample_safety_12g[

    "Induced_FN_Rate"

] = np.where(



    sample_safety_12g[

        "Affected_Phishing_Count"

    ] > 0,



    (

        sample_safety_12g[

            "Induced_FN_Flips"

        ]



        /



        sample_safety_12g[

            "Affected_Phishing_Count"

        ]

    ),



    0.0

)





```

## Results: FN-safety screening

I tested the changed features and synthetic email variants using five training-only validation folds. The two safety checks were designed to identify changes that might turn a correctly detected phishing email into a false negative.

The following tables and figures show the **recorded results from the original experiment**. They are included here so the findings are visible on GitHub without running the notebook. The code above remains the source for reproducing and checking these results.

**Table 6.1 — Feature-level safety results**

| Result | Number of features | Share of tested features |
|:--|--:|--:|
| Passed the FN-safety test | **62** | 66.67% |
| Flagged for FN risk | **31** | 33.33% |
| **Total changed features tested** | **93** | **100.00%** |

A feature passed this screening when the tested reduction did not create any additional false-negative predictions in the selected validation examples. This does not guarantee that the feature is safe in every possible email.


```python
# Recreate Figure 6.1 from the screening results above.
import matplotlib.pyplot as plt
assert len(token_safety_12g) == 93, "The feature count differs from the recorded experiment."
assert safe_token_count_12g == 62 and risky_token_count_12g == 31, "Feature safety counts differ from the recorded experiment."
fig, ax = plt.subplots(figsize=(8, 4.3))
labels = ["Passed", "Flagged for FN risk"]
values = [safe_token_count_12g, risky_token_count_12g]
bars = ax.bar(labels, values, color=["#2c617d", "#b27c52"], width=0.49)
ax.set(title="Figure 6.1 — Changed features after FN-safety screening", ylabel="Number of changed features", ylim=(0, max(values) * 1.23))
ax.spines[["top", "right"]].set_visible(False)
for bar, value in zip(bars, values):
    ax.text(bar.get_x() + bar.get_width()/2, value + 1, str(value), ha="center", fontweight="bold")
plt.tight_layout()
plt.show()
```

**Table 6.2 — Safety screening of synthetic legitimate emails**

| Result | Number of variants | Share of approved variants |
|:--|--:|--:|
| Passed both safety tests | **9** | 26.47% |
| Rejected after screening | **25** | 73.53% |
| **Total variants screened** | **34** | **100.00%** |

I retained **9 of the 34** approved synthetic variants because they passed both the individual-feature test and the combined-pattern test. The remaining **25** did not meet the full safety rule. No synthetic variants were added to model training in this notebook.

**Table 6.3 — Screening procedure**

| Stage | What I checked | Rule used |
|:--|:--|:--|
| Individual-feature screening | The effect of reducing each changed TF-IDF feature in correctly detected phishing examples | No additional false negatives |
| Combined-pattern screening | The effect of applying the full set of feature changes linked to a synthetic variant | No additional false negatives |
| Final selection | Whether each synthetic variant passed both tests | Both tests must pass |

*Note:* These are training-only screening results. They do not establish the final test-set performance of a retrained classifier.


```python
# Recreate Figure 6.2 from the screening results above.
import matplotlib.pyplot as plt
assert safe_samples_count_12g == 9 and risky_samples_count_12g == 25, "Variant counts differ from the recorded experiment."
fig, ax = plt.subplots(figsize=(8, 4.3))
labels = ["Passed both tests", "Rejected"]
values = [safe_samples_count_12g, risky_samples_count_12g]
bars = ax.bar(labels, values, color=["#2c617d", "#b27c52"], width=0.49)
ax.set(title="Figure 6.2 — Synthetic email variants after FN-safety screening", ylabel="Number of synthetic variants", ylim=(0, max(values) * 1.23))
ax.spines[["top", "right"]].set_visible(False)
for bar, value in zip(bars, values):
    ax.text(bar.get_x() + bar.get_width()/2, value + 0.5, str(value), ha="center", fontweight="bold")
plt.tight_layout()
plt.show()
```

### Interpretation

The screening removed most of the synthetic variants before retraining. I kept only the nine variants that met the strict safety rule in the training-only checks. This is an important precaution because my aim is to reduce false positives without creating more false negatives.

These results describe **screening**, not final model performance. I still need to evaluate the retrained model separately before claiming any improvement.
