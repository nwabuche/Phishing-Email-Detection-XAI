# 08 — DistilBERT: SHAP-Guided Corrective Example Selection

**Research project:** Explainability-Guided Feature Analysis for Reducing False Positives in Phishing Email Detection Without Increasing False Negatives

This notebook presents the selection of synthetic legitimate emails for additional DistilBERT training. SHAP explanations of false-positive predictions guided candidate generation. The candidates were then reviewed and screened to produce a smaller set of corrective examples. This stage concerns **training-example selection**, not a change to the test set or a new model evaluation.


## 1. Research objective

The official DistilBERT baseline produced **31 false positives** (legitimate emails classified as phishing) and **22 false negatives** (phishing emails classified as legitimate). The aim of this experiment was to develop legitimate training examples that reflected the language patterns associated with false-positive predictions.

SHAP explanations were used to examine the features influencing these predictions. Candidate emails were generated from the identified contexts and screened before inclusion in a later fine-tuning experiment. The wider research objective was to reduce false positives without increasing false negatives.


## 2. Dataset and baseline reference

**Table 8.1. Dataset and baseline error summary**

| Dataset or classification measure | Number of emails |
|:---|---:|
| Cleaned dataset | 53,577 |
| Original training set | 42,861 |
| Fixed test set | 10,716 |
| Legitimate emails in the training set | 18,991 |
| Phishing emails in the training set | 23,870 |
| Baseline false positives | **31** |
| Baseline false negatives | **22** |

The experiment used the saved official DistilBERT baseline. Class **0** represented legitimate emails, and class **1** represented phishing emails. In the terminology used throughout this project, **TP** refers to correctly classified legitimate emails and **TN** refers to correctly classified phishing emails.

The synthetic emails were prepared as potential additions to the training data. The fixed test set was not modified.


## 3. SHAP-guided candidate generation

SHAP explanations were examined for the **31 baseline false-positive emails**. The analysis considered the words that contributed to phishing predictions and the legitimate context in which those words appeared.

Using this evidence, the experiment generated variations of legitimate emails intended to address the observed classification errors. The generated emails were candidates for further screening; they were not automatically accepted for training.

**Table 8.2. Candidate generation summary**

| Measure | Number |
|:---|---:|
| Baseline false-positive emails examined | 31 |
| False-positive contexts used to generate candidates | 24 |
| Synthetic legitimate candidates generated | **129** |

No additional model training took place during candidate generation.


## 4. Candidate verification and screening

The generated emails were checked using the recorded selection procedure. **128** of the **129** generated candidates were recovered for verification. Screening then retained **26** candidates representing **8** false-positive contexts. A final review selected **21** legitimate examples representing **6** contexts.

**Table 8.3. Corrective-example selection summary**

| Selection stage | Candidate emails | False-positive contexts |
|:---|---:|---:|
| Initial candidate generation | 129 | 24 |
| Candidates recovered for verification | 128 | Not recorded |
| Candidates retained after screening | 26 | 8 |
| **Final selected examples** | **21** | **6** |

The recorded counts show that one fewer candidate was available at verification than at generation. The available summary does not establish the reason for that difference.

The final 21 examples were **synthetic legitimate training examples**, not additional test emails. Selecting a smaller group allowed the subsequent fine-tuning experiment to focus on candidates that passed the recorded checks.


## 5. Final corrective-example dataset

The screening process produced a focused set of legitimate examples for later model fine-tuning. The original training records remained in place during this selection stage; the selected examples were prepared separately.

**Table 8.4. Final selection summary**

| Property | Final selection |
|:---|:---|
| Example type | Synthetic legitimate emails |
| Class label | 0 (legitimate) |
| Number of selected examples | **21** |
| Distinct false-positive source contexts | **6** |
| Original training records replaced | No |
| Additional model training at this stage | No |

The effect of adding these examples to the training data is examined in the subsequent fine-tuning experiment. The selection counts alone do not establish an improvement in classification performance.


## 6. Numerical consistency checks

The following code checks the recorded dataset totals and the order of the candidate-selection counts. It does not repeat SHAP analysis or candidate generation.


```python
training_legitimate = 18_991
training_phishing = 23_870
training_total = 42_861
cleaned_total = 53_577
test_total = 10_716

candidate_counts = {
    "Generated candidates": 129,
    "Recovered candidates": 128,
    "Candidates retained after screening": 26,
    "Final selected examples": 21,
}
final_source_contexts = 6

assert training_legitimate + training_phishing == training_total
assert training_total + test_total == cleaned_total
assert list(candidate_counts.values()) == sorted(candidate_counts.values(), reverse=True)
assert 0 < final_source_contexts <= candidate_counts["Final selected examples"]

print("Dataset totals and candidate-selection counts are consistent.")
for stage, count in candidate_counts.items():
    print(f"{stage}: {count:,}")
print(f"Distinct contexts in final selection: {final_source_contexts}")
```

## 7. Interpretation of findings

The SHAP-guided procedure generated **129** candidate legitimate emails from **24** false-positive contexts. Following verification and screening, **21** examples from **6** contexts were selected for additional training.

This result shows how the initial candidate pool was narrowed to a smaller set of corrective examples. The selected emails were intended to help DistilBERT distinguish legitimate messages from phishing messages when they contain similar language patterns. Their effect on false positives and false negatives was assessed during the later model fine-tuning and evaluation.


## 8. Conclusion

This experiment used SHAP-guided error analysis to prepare **21 synthetic legitimate emails** from **6 false-positive contexts**. The examples were selected through successive verification and screening stages and were reserved for a subsequent DistilBERT fine-tuning experiment.

The selection procedure provided targeted training material for investigating whether false positives could be reduced while maintaining phishing detection performance.
