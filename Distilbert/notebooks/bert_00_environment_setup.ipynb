## 2. Check the original experiment's Python dependencies

The shared DistilBERT stages use PyTorch and Hugging Face Transformers for model inference/fine-tuning, SHAP for explanations, and pandas, NumPy, scikit-learn, and Matplotlib for data processing and evaluation. This check **does not install or upgrade** any packages; it only reports what is available in the current environment.

The recorded experiment also uses Google Colab/Google Drive paths. `google.colab` is checked separately because it is specific to the Colab runtime.

```python
from importlib.util import find_spec
from importlib.metadata import version, PackageNotFoundError

packages = [
    ("numpy", "numpy"),
    ("pandas", "pandas"),
    ("torch", "torch"),
    ("transformers", "transformers"),
    ("shap", "shap"),
    ("sklearn", "scikit-learn"),
    ("matplotlib", "matplotlib"),
]

missing = []
for module_name, distribution_name in packages:
    available = find_spec(module_name) is not None
    if available:
        try:
            installed_version = version(distribution_name)
        except PackageNotFoundError:
            installed_version = "version unavailable"
        print(f"{distribution_name:16} AVAILABLE  {installed_version}")
    else:
        print(f"{distribution_name:16} MISSING")
        missing.append(distribution_name)

print("\nDependency availability check:", "PASSED" if not missing else "INCOMPLETE")
if missing:
    print("Missing packages:", ", ".join(missing))
```

## 3. Check PyTorch and GPU availability

The original DistilBERT training/evaluation was carried out using a **Tesla T4 GPU in Google Colab**. This check reports whether CUDA is available **now**. A different GPU or a CPU-only environment does not change the historical experiment, but it may affect whether later notebooks can be executed efficiently.

If PyTorch is not installed, this check reports that it is unavailable rather than stopping the notebook.


```python
from importlib.util import find_spec

if find_spec("torch") is None:
    print("PyTorch: NOT AVAILABLE")
    print("GPU verification skipped because PyTorch is not installed.")
else:
    try:
        import torch

        print("PyTorch version:", torch.__version__)
        print("CUDA available:", torch.cuda.is_available())
        if torch.cuda.is_available():
            print("CUDA device:", torch.cuda.get_device_name(0))
        else:
            print("No CUDA GPU detected in this runtime.")
    except (ImportError, OSError) as exc:
        print("PyTorch could not be loaded:", exc)
        print("GPU verification could not be completed.")
```
