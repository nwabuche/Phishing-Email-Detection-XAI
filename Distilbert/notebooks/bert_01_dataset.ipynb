## 2. Recorded text preparation

The original DistilBERT preparation was described as conservative text cleaning. The recorded operations were:

1. Convert email text to lowercase.
2. Remove URL patterns.
3. Remove email-address patterns.
4. Remove non-alphanumeric characters.
5. Exclude cleaned texts shorter than **20 characters**.

## 4. Recorded training and test split

The original DistilBERT experiment used an **80% training / 20% test split**, with **stratification by label** and **`random_state=42`**.

| Subset | Legitimate (0) | Phishing (1) | Total |
|---|---:|---:|---:|
| Training | 18,991 | 23,870 | **42,861** |
| Test | 4,748 | 5,968 | **10,716** |
| **Total** | **23,739** | **29,838** | **53,577** |

The test set is the same recorded test population used for the official DistilBERT baseline and later Step 19G evaluation.

## 5. Consistency checks for the recorded counts

The following code verifies **arithmetic consistency of the published counts only**. It does not download data, clean emails, regenerate the split, or verify individual email membership.

```python
original_count = 53_668
cleaned_count = 53_577
removed_count = 91

train_legitimate = 18_991
train_phishing = 23_870
test_legitimate = 4_748
test_phishing = 5_968

assert original_count - cleaned_count == removed_count
assert train_legitimate + train_phishing == 42_861
assert test_legitimate + test_phishing == 10_716
assert train_legitimate + test_legitimate == 23_739
assert train_phishing + test_phishing == 29_838
assert 42_861 + 10_716 == cleaned_count

print("Recorded dataset-count consistency: PASS")
print(f"Original emails: {original_count:,}")
print(f"Cleaned emails: {cleaned_count:,}")
print(f"Training emails: {train_legitimate + train_phishing:,}")
print(f"Test emails: {test_legitimate + test_phishing:,}")
```
