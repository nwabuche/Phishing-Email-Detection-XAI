# DistilBERT — SHAP-Based Error Analysis

**Research project:** Explainability-Guided Feature Analysis for Reducing False Positives in Phishing Email Detection Without Increasing False Negatives  
**Model:** DistilBERT  
**Focus:** Interpretation of false-positive predictions using SHAP.

## 1. Research objective

This analysis used SHapley Additive exPlanations (SHAP) to investigate why the DistilBERT classifier identified some legitimate emails as phishing. The main objective was to examine the text features associated with false-positive predictions and use the findings to guide later experiments. False negatives were also monitored because reducing phishing emails incorrectly classified as legitimate remained an important requirement.

The dataset used two classes: **0 = legitimate** and **1 = phishing**. The terminology used in this project is **TP = correctly classified legitimate**, **TN = correctly classified phishing**, **FP = legitimate classified as phishing**, and **FN = phishing classified as legitimate**.

## 2. Baseline classification errors

**Table 3.1. Classification errors in the official DistilBERT baseline**

| Error category | Description | Number of emails |
|:---|:---|---:|
| False positive (FP) | Legitimate email classified as phishing | 31 |
| False negative (FN) | Phishing email classified as legitimate | 22 |
| **Total** | **All incorrectly classified emails** | **53** |

The baseline produced **31 false positives** and **22 false negatives**. These errors provided the starting point for the explainability analysis.

## 3. SHAP explanation method

SHAP was applied to the text-classification model to examine the contribution of individual tokens to predictions for the phishing class. The analysis focused on the baseline false-positive emails and examined recurring tokens across their explanations.

**Table 3.2. Configuration of the SHAP analysis**

| Component | Experimental setting |
|:---|:---|
| Explanation method | SHapley Additive exPlanations (SHAP) |
| Explainer | `shap.PartitionExplainer` |
| Input masker | SHAP text masker |
| Explained output | Phishing class (class 1) |
| Main analysis group | Baseline false-positive emails |
| Token analysis | Recurring tokens across false-positive explanations |
| Aggregation threshold | Token frequency of at least 3 |
| Selected feature set | Top 20 tokens from the recorded analysis |

SHAP values indicate how tokens contribute to a particular model prediction. The results were used to study prediction patterns rather than to assume that any single token caused an error.

## 4. Error-analysis procedure

The baseline errors were identified first. SHAP explanations were then examined to identify tokens that appeared repeatedly in false-positive predictions. A set of 20 tokens was selected and saved as a fixed reference for later experiments.

**Table 3.3. Main activities in the SHAP error analysis**

| Activity | Description | Research purpose |
|:---|:---|:---|
| Baseline evaluation | Identify 31 false positives and 22 false negatives | Establish the initial classification errors |
| SHAP explanation | Examine token contributions for false-positive emails | Understand the model's phishing predictions |
| Feature selection | Identify the 20 selected tokens | Summarize recurring features for further analysis |
| Feature preservation | Save the selected token list as a JSON file | Keep the feature reference consistent |
| Follow-up experiments | Examine targeted changes and compare error counts | Evaluate false-positive reduction while monitoring false negatives |

The selected token list was retained as an experimental record so that later analyses could refer to the same features.

## 5. Saved SHAP feature list

The selected tokens were saved in a JSON file in the original project directory. The code below checks whether this file is available, reads its structure, and calculates a SHA-256 checksum. This provides a way to inspect the saved feature record without modifying it.

**Table 3.4. Saved feature record**

| Property | Description |
|:---|:---|
| Content | Selected Top-20 SHAP tokens |
| Format | JSON |
| Storage | Original DistilBERT project directory in Google Drive |
| Integrity check | SHA-256 checksum |

The storage path in the code retains its original experimental filename to locate the saved file. These internal names are not used as research-stage labels in the discussion.

```python
from pathlib import Path
import hashlib
import json

# The original experiment's directory and file names are retained for access.
project_root = Path(
    "/content/drive/MyDrive/"
    "DistilBERT_SHAP_Guided_Explainability_Based_Fine_Tuning"
)
feature_file = (
    project_root
    / "Step10D_Permanent_Top20_SHAP_Token_LOCK"
    / "DistilBERT_Step10D_Locked_Top20_Tokens.json"
)

if not feature_file.is_file():
    print("Saved SHAP feature file was not found at the specified path.")
    print("File path:", feature_file)
else:
    raw = feature_file.read_bytes()
    saved_features = json.loads(raw.decode("utf-8"))
    checksum = hashlib.sha256(raw).hexdigest()
    print("Saved feature file:", feature_file.name)
    print("SHA-256:", checksum)
    if isinstance(saved_features, dict):
        print("JSON fields:", sorted(saved_features))
    elif isinstance(saved_features, list):
        print("Number of saved entries:", len(saved_features))
    else:
        print("JSON content type:", type(saved_features).__name__)
```

The SHAP analysis examined how individual text tokens contributed to phishing predictions for legitimate emails that the baseline model classified incorrectly. Repeated token patterns were used to identify a selected set of features for further investigation.

The selected features were saved to maintain a consistent reference across subsequent experiments. This connected the explainability analysis with the development of targeted changes to the classifier, while keeping both false-positive and false-negative results visible.
