## 1. Research objective

The official DistilBERT baseline was examined using SHapley Additive exPlanations (SHAP) to investigate why legitimate emails were incorrectly classified as phishing. The class mapping was **0 = legitimate** and **1 = phishing**. The original project terminology is retained: **TP = correctly classified legitimate**, **TN = correctly classified phishing**, **FP = legitimate classified as phishing**, and **FN = phishing classified as legitimate**.

## 5. Inspect the original permanent Top-20 token lock (optional)

This **read-only** cell checks for the original JSON file saved by Step 10D in Google Drive. It does not generate a replacement token list. Run it only in an environment with the original project Drive mounted.

**Table 3.2. Expected original SHAP lock artifact**

| **Item** | **Original record** |
|:--|:--|
| Stage | Step 10D — permanent Top-20 SHAP token lock |
| Filename | `DistilBERT_Step10D_Locked_Top20_Tokens.json` |
| Storage | Google Drive, under the project's Step 10D lock directory |
| Integrity mechanism | SHA-256 in the original Step 10D workflow |

```python
from pathlib import Path
import hashlib
import json

project_root = Path(
    "/content/drive/MyDrive/"
    "DistilBERT_SHAP_Guided_Explainability_Based_Fine_Tuning"
)
lock_file = (
    project_root
    / "Step10D_Permanent_Top20_SHAP_Token_LOCK"
    / "DistilBERT_Step10D_Locked_Top20_Tokens.json"
)

if not lock_file.is_file():
    print("Original Step 10D lock not found; no tokens are reconstructed.")
    print("Expected location:", lock_file)
else:
    raw = lock_file.read_bytes()
    payload = json.loads(raw.decode("utf-8"))
    digest = hashlib.sha256(raw).hexdigest()
    print("Original lock file located:", lock_file)
    print("SHA-256 of the available file:", digest)
    print("JSON structure:", type(payload).__name__)
    if isinstance(payload, dict):
        print("Available JSON keys:", sorted(payload.keys()))
    elif isinstance(payload, list):
        print("Number of top-level list entries:", len(payload))
    print("Compare this checksum with the original Step 10D recorded checksum before relying on the artifact.")
```
